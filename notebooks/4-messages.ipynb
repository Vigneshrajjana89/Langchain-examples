# Messages in LangChain

This notebook demonstrates using Chat Models and different types of Messages (`HumanMessage`, `SystemMessage`, `AIMessage`).

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

load_dotenv()

# Always use a valid active model ID for Groq:
# e.g. 'llama-3.3-70b-versatile', 'llama-3.1-8b-instant', or 'qwen/qwen3.6-27b'
model = ChatGroq(model="llama-3.3-70b-versatile")
```

```python
# Simple prompt invocation
response = model.invoke("Please tell what is artificial intelligence")
print(response.content)
```

```python
# Using structured message types
messages = [
    SystemMessage(content="You are a helpful and concise tech tutor."),
    HumanMessage(content="What is artificial intelligence?")
]

response = model.invoke(messages)
print(response.content)
```
